```python
import pandas as pd
import numpy as np
import glob
import os
```

```python
def Yearly_sales(df):
    df.dropna(inplace=True)
    df.rename(columns={"Price Each":"Price"},inplace=True)
    df.rename(columns={"Quantity Ordered":"Quantity"},inplace=True)
    df.rename(columns={"Purchase Address":"Address"},inplace=True)
    df=df.drop_duplicates()
    df=df[df["Price"] != "Price Each"]
    df["Price"]=df["Price"].astype(float)
    df["Quantity"]=df["Quantity"].astype(int)
    df["Order Date"] = pd.to_datetime(df["Order Date"])
    df["Total"]= df["Quantity"] * df["Price"]
    df['Day'] = df['Order Date'].dt.day_name()
    df['Month'] = df['Order Date'].dt.month_name()

    def address(x):
     x = x.split(',')[1]
     return x
    df['City'] = df['Address'].apply(lambda x: address(x))
    return df
```
